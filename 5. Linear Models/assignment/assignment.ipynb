```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score

print("-----------Part 1----------")
df = pd.read_csv("./data/Q1_clean.csv")
df.columns = df.columns.str.strip()
stats = df.groupby('Neighbourhood')[['Price', 'Review Scores Rating']].mean()
print(stats.sort_values('Price', ascending=False))
print("The most expensive bourough seems to be Manhattan, with an average price per night at $183.7, whereas the next one is Staten Island, at $146.2")
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6)) # Had to ask Gemini for help setting this up as it's been a while.
sns.kdeplot(data=df, x='Price', hue='Neighbourhood', fill=True, ax=ax1)
ax1.set_title('Kernel Density Estimate of Price by Neighbourhood')
ax1.set_xlabel('Price ($)')
sns.kdeplot(x=np.log(df['Price']), hue=df['Neighbourhood'], fill=True, ax=ax2)
ax2.set_title('Kernel Density Estimate of Log(Price) by Neighbourhood')
ax2.set_xlabel('Natural Log Price}')
plt.tight_layout()
plt.show()
print("The non-log kernel shows a tall skyscraper at the actual values due to outliers, whereas the log(price) shows a centered data structure")

print("\n"+"-----------Part 2----------"+"\n")
X = pd.get_dummies(df["Neighbourhood"])
y = df["Price"]
model = LinearRegression(fit_intercept=False)
model.fit(X, y)
coef_df = pd.DataFrame({'Borough': X.columns, 'Coefficient': model.coef_})
print(coef_df)
print("They are the exact same as the price averages in the first part. Because there is no intercept, the parameter is just changing it to true when correct, which just causes it to instantly go to the average price it was, because without an intercept and only the same parameters, it has to come directly from that one, making it just end up being the average.")

print("\n"+"-----------Part 3----------"+"\n")
X2 = pd.get_dummies(df["Neighbourhood"],drop_first=True)
y = df["Price"]
model1 = LinearRegression(fit_intercept=True)
model1.fit(X2, y)
coef_df = pd.DataFrame({'Borough': X2.columns, 'Coefficient': model1.coef_})
print(coef_df)
print("Now the values are slighty lower, this is because it's adding an intercept to the equation that all of the different values share. This can be calculated by doing 183.664286-108.387789 which gives 75.27 as the intercept. This is also equal to the value of the lowest Neighbourhood, or Bronx. For the creation, I had to remove one set because the intercept would of remove one that otherwise would of been included.")

print("\n"+"-----------Part 4----------"+"\n")
X3 = pd.get_dummies(df[['Review Scores Rating', 'Neighbourhood']], drop_first=True)
y= df["Price"]
X_train, X_test, y_train, y_test = train_test_split(X3, y, test_size=0.2, random_state=42) # Had to get a little reminder from gemini how to set up the testing, as well as looking back at my old assignments.
model3 = LinearRegression()
model3.fit(X_train, y_train)
y_pred = model3.predict(X_test)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)
print(f"R-squared: {r2:.4f}")
print(f"RMSE: {rmse:.2f}")
print(f"Review Rating Coeff: {model3.coef_[0]:.4f}")
print("The r squared value is really low, showing a low correlation directly between reviews, neighbourhood, with price")

print("\n"+"-----------Part 5----------"+"\n")
X4 = pd.get_dummies(df[['Review Scores Rating', 'Neighbourhood',"Property Type"]], drop_first=True)
y= df["Price"]
X_train1, X_test1, y_train1, y_test1 = train_test_split(X4, y, test_size=0.2, random_state=43)
model4 = LinearRegression()
model4.fit(X_train1, y_train1)
y_pred1 = model4.predict(X_test1)
rmse1 = np.sqrt(mean_squared_error(y_test1, y_pred1))
r21 = r2_score(y_test1, y_pred1)
print(f"R-squared: {r21:.4f}")
print(f"RMSE: {rmse1:.2f}")
print(f"Review Rating Coeff: {model4.coef_[0]:.4f}")
print("The r squared value is really low, showing a low correlation directly between reviews, neighbourhood, property type with price")
coef_df = pd.DataFrame({'Feature': X4.columns, 'Coefficient': model4.coef_})
property_rank = coef_df[coef_df['Feature'].str.contains('Property Type')].sort_values(by='Coefficient', ascending=False)
print(property_rank.head(1))
print("Is the most expensive property")

print("\n"+"-----------Part 6----------"+"\n")
print("Because property type was added, it changed the importance of each variable and determined how much it mattered. In part 4, it only looked at where it was and the reviews, and therefore htey both got big ratings. But, in part 5, it was able to also look at property type, and because property type made a bigger impact, it slighty lowered the parameter/weight on it.")


print("\n" + "Currently do not have time for part 7")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score

print("-----------Part 1----------")
df = pd.read_csv("./data/cars_hw.csv")
df['Log_Price'] = np.log(df['Price'])
df['Log_Miles'] = np.log(df['Mileage_Run'])
df["New_Transmission"] = df['Transmission'].str.contains('Speed')
print("Done")

print("\n"+"-----------Part 2----------"+"\n")
price_summary = df['Price'].describe()
print(price_summary)
plt.figure(figsize=(10, 6))
sns.kdeplot(df['Price'], fill=True, color='royalblue')
plt.title('Distribution of Car Prices')
plt.xlabel('Price ($)')
plt.ylabel('Density')
plt.show()

top_brands = df['Make'].value_counts().nlargest(5).index # This and a little bit on the summary I had to ask gemini to assist me on.
df_top = df[df['Make'].isin(top_brands)]
brand_summary = df.groupby('Make')['Price'].describe().sort_values(by='mean', ascending=False)
print(brand_summary.head(10))
plt.figure(figsize=(12, 6))
sns.kdeplot(data=df_top, x='Price', hue='Make', fill=True, alpha=0.3)
plt.title('Price Distribution by Brand (Top 5 Most Frequent Brands)')
plt.xlabel('Price ($)')
plt.ylabel('Density')
plt.show()
print("The most expensive car brand was MG Motors with a value of around 1.87 million dollars, tho Skoda had a higher maximum value, it's average was lower. The other most expensive car brands are shown in the mean column under the first graph. Prices can be skwewed due to a few select cars being more fancy than others, causing an inflated maximum value, but a lower average that's still being skewed by these values.")

print("\n"+"-----------Part 3----------"+"\n")
y = df["Price"]
X = df.select_dtypes(include=['float64', 'int64']).drop(columns=['Price'])
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=44)
print(f"Original data shape: {df.shape}")
print(f"Training set (X_train): {X_train.shape}")
print(f"Testing set (X_test): {X_test.shape}")

print("\n"+"-----------Part 4----------"+"\n")
X_num = df.select_dtypes(include=[np.number]).drop(columns=['Price'])
y = df['Price']
X_train2, X_test2, y_train2, y_test2 = train_test_split(X_num, y, test_size=0.2, random_state=14)
model2 = LinearRegression()
model2.fit(X_train2, y_train2)
train_preds2 = model2.predict(X_train2)
test_preds2 = model2.predict(X_test2)
print(f"Model 1 Test R2: {r2_score(y_test2, test_preds2)}")
print(f"Model 1 Test RMSE: {np.sqrt(mean_squared_error(y_test2, test_preds2))}")

df_cat = pd.get_dummies(df.select_dtypes(include=['object', 'string']), drop_first=True)
y = df['Price']
X_train3, X_test3, y_train3, y_test3 = train_test_split(df_cat, y, test_size=0.2, random_state=14)
model3 = LinearRegression()
model3.fit(X_train3, y_train3)
train_preds3 = model3.predict(X_train3)
test_preds3 = model3.predict(X_test3)
print(f"Model 2 Test R2: {r2_score(y_test3, test_preds3)}")
print(f"Model 2 Test RMSE: {np.sqrt(mean_squared_error(y_test3, test_preds3))}")

X_joint = pd.concat([X_num, df_cat], axis=1)
X_train_j, X_test_j, y_trainj, y_testj = train_test_split(X_joint, y, test_size=0.2, random_state=33)
model4 = LinearRegression()
model4.fit(X_train_j, y_trainj)
test_preds4 = model4.predict(X_test_j)
print(f"Model 3 Test R2: {r2_score(y_testj, test_preds4)}")
print(f"Model 3 Test RMSE: {np.sqrt(mean_squared_error(y_testj, test_preds4))}")
print("The categorial values only seem to have the worst correlation, at 0.75, whereas the numerical has a 0.9. The joint one, combining the 2 different training sets, did come up with a larger correlation at 0.93, meaning it did perform better.")

print("\n"+"-----------Part 5----------"+"\n")

from sklearn.preprocessing import PolynomialFeatures
results = []

for d in range(1,5): # Had Gemini help me out creating this function.
    poly = PolynomialFeatures(degree=d)
    X_poly = poly.fit_transform(X_num)
    X_train_p, X_test_p, y_train_p, y_test_p = train_test_split(X_poly, y, test_size=0.2, random_state=14)
    model_p = LinearRegression()
    model_p.fit(X_train_p, y_train_p)
    test_preds_p = model_p.predict(X_test_p)
    r2 = r2_score(y_test_p, test_preds_p)
    rmse = np.sqrt(mean_squared_error(y_test_p, test_preds_p))
    results.append((d, r2, rmse))
    print(f"Degree {d} - R2: {r2:.4f}, RMSE: {rmse:.2f}")
print("The model slowly improves throughout the first 3 degrees, but after that it starts going down, due to likely overfitting the data. Excluding the first point, this is also shown by the deep increase in RMSE, versus the decrease before it. After the first 20 tests, it actually never becomes negatiev and instead just keeps sitting around a low 2-3%. The best degree, which is 3, seems to outperform the best model, the joint model, in the last problem by a decent bit of 5 percent, meaning it's more accurate.")

print("\n"+"-----------Part 6----------"+"\n")
poly_best = PolynomialFeatures(degree=3)
X_poly_train = poly_best.fit_transform(X_train2)
X_poly_test = poly_best.transform(X_test2)
model_best = LinearRegression()
model_best.fit(X_poly_train, y_train2)
predictions = model_best.predict(X_poly_test)
plt.figure(figsize=(8, 8))
plt.scatter(y_test2, predictions, alpha=0.5, color='seagreen')
plt.plot([y_test2.min(), y_test2.max()], [y_test2.min(), y_test2.max()], 'r--', lw=2)
plt.xlabel('Actual Price')
plt.ylabel('Predicted Price')
plt.title('Actual vs. Predicted Prices (Degree 3 Polynomial)')
plt.show()
residuals = y_test2 - predictions
plt.figure(figsize=(10, 6))
sns.kdeplot(residuals, fill=True, color='crimson')
plt.axvline(0, color='black', linestyle='--')
plt.title('Kernel Density Plot of Residuals (Degree 3 Model)')
plt.xlabel('Residual (Actual Price - Predicted Price)')
plt.ylabel('Density')
plt.show()
print("Yes, the predicted values do line up really well with the line. The residual is a bell curve centered around zero, showing this is the best fit for the data we could of done. It's slight to the right, which could be a little weakness, and it only uses the numerical values, but it does have a strength as the correlation is really high and almost accurate.")
```
